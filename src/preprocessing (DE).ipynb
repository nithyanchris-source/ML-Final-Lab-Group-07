```python
import pandas as pd

# Assuming the file is in your Google Drive, in the root 'MyDrive' folder.
df = pd.read_csv('/content/imdb_combined.csv')

# Display the first few rows of the DataFrame to verify
print(df.head())
```

```python
# Remove the 'original_split' column (the third column)
df = df.drop(columns=['original_split'], errors = 'ignore')

# Display the first few rows of the updated DataFrame to verify
print(df.head())
```

### Data Preprocessing Steps

1.  **Check for Missing Values**: Identify and handle any null entries.
2.  **Text Cleaning**: Clean the 'review' column by converting text to lowercase and removing punctuation. This is a basic step to prepare text data for further analysis.

```python
print('Checking for missing values:')
print(df.isnull().sum())

# As there are no missing values based on the output, we can proceed with text cleaning.
```

```python
import re

def clean_text(text):
    text = text.lower()  # Convert to lowercase
    text = re.sub(r'[^a-z0-9\s]', '', text)  # Remove punctuation and special characters, keep alphanumeric and spaces
    return text

print('\nApplying text cleaning to the \'review\' column...')
df['review'] = df['review'].apply(clean_text)

print('\nPreprocessing complete. Displaying info and first few rows of the cleaned data:')
df.info()
print(df.head())
```

### Text Tokenization

Tokenization is the process of breaking text into smaller units called tokens (e.g., words, phrases, symbols). This is a fundamental step in most NLP tasks.

```python
import nltk
nltk.download('punkt') # Download the necessary tokenizer data
nltk.download('punkt_tab') # Download 'punkt_tab' as suggested by the error

print('\nTokenizing the \'review\' column...')
df['review_tokens'] = df['review'].apply(nltk.word_tokenize)

print('\nTokenization complete. Displaying the first few rows with tokens:')
print(df[['review', 'review_tokens']].head())
```

### Stop Word Removal

Stop words are words that are filtered out before or after processing of natural language data. Removing them helps in focusing on more important words for analysis and reducing dimensionality.

```python
from nltk.corpus import stopwords
nltk.download('stopwords') # Download the necessary stop words data

stop_words = set(stopwords.words('english'))

def remove_stopwords(tokens):
    return [word for word in tokens if word not in stop_words]

print('\nRemoving stop words from the tokenized reviews...')
df['review_tokens_cleaned'] = df['review_tokens'].apply(remove_stopwords)

print('\nStop word removal complete. Displaying the first few rows with cleaned tokens:')
print(df[['review', 'review_tokens', 'review_tokens_cleaned']].head())
```

### Data Quality Report

This section provides a summary of the dataset's quality, including sentiment distribution, missing values, duplicates, and uniqueness of reviews.

```python
print('--- Sentiment Distribution ---')
print(df['sentiment'].value_counts())

print('\n--- Missing Values ---')
print(df.isnull().sum())

print('\n--- Duplicate Reviews ---')
duplicate_reviews = df.duplicated(subset=['review']).sum()
print(f'Number of duplicate reviews: {duplicate_reviews}')

print('\n--- Unique Reviews ---')
unique_reviews_count = df['review'].nunique()
print(f'Number of unique reviews: {unique_reviews_count}')

print('\n--- DataFrame Information ---')
df.info()
```

### Removing Duplicate Reviews

Based on the quality report, there are `419` duplicate reviews. These will now be removed to ensure that each review entry is unique.

```python
# Remove duplicate reviews based on the 'review' column
df.drop_duplicates(subset=['review'], inplace=True)

print('Duplicate reviews removed. Verifying...')

# Re-check for duplicate reviews
duplicate_reviews_after_removal = df.duplicated(subset=['review']).sum()
print(f'Number of duplicate reviews after removal: {duplicate_reviews_after_removal}')

# Display the new shape of the DataFrame
print(f'New DataFrame shape: {df.shape}')
```

### Stop Word Removal

Stop words are words that are filtered out before or after processing of natural language data. Removing them helps in focusing on more important words for analysis and reducing dimensionality.

```python
from nltk.corpus import stopwords
nltk.download('stopwords') # Download the necessary stop words data

stop_words = set(stopwords.words('english'))

def remove_stopwords(tokens):
    return [word for word in tokens if word not in stop_words]

print('\nRemoving stop words from the tokenized reviews...')
df['review_tokens_cleaned'] = df['review_tokens'].apply(remove_stopwords)

print('\nStop word removal complete. Displaying the first few rows with cleaned tokens:')
print(df[['review', 'review_tokens', 'review_tokens_cleaned']].head())
```

```python
from sklearn.model_selection import train_test_split

# Define features (X) and target (y)
X = df['review_tokens_cleaned'] # Using the cleaned, tokenized reviews
y = df['sentiment']

# Split the data into training and testing sets
# Stratify by 'sentiment' to ensure an even distribution of classes in both sets
# random_state for reproducibility
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

print(f'Training set size: {len(X_train)} samples')
print(f'Testing set size: {len(X_test)} samples')

print('\nSentiment distribution in training set:')
print(y_train.value_counts(normalize=True))

print('\nSentiment distribution in testing set:')
print(y_test.value_counts(normalize=True))
```

```python
df = df.drop(
    columns=['review_tokens'],
    errors='ignore'
)
```

### Detailed Train/Test Leakage Checking

This step checks for any overlap in the unique reviews between the training and testing datasets. Even after deduplication of the original dataset, accidental leakage can occur if the split is not carefully performed, leading to an over-optimistic evaluation of the model.

```python
# Convert token lists back to strings for comparison
X_train_str = X_train.apply(lambda x: ' '.join(x))
X_test_str = X_test.apply(lambda x: ' '.join(x))

# Find common reviews between train and test sets
common_reviews = set(X_train_str.unique()).intersection(set(X_test_str.unique()))
leakage_count = len(common_reviews)

print('--- Checking for Train/Test Leakage ---')
if leakage_count == 0:
    print('No leakage detected between training and testing sets.')
else:
    print(f'WARNING: {leakage_count} common reviews found between training and testing sets. This indicates potential data leakage.')
    # Optionally, print some of the leaked reviews
    # print('Leaked reviews examples:')
    # for i, review in enumerate(list(common_reviews)[:5]): # Print up to 5 examples
    #     print(f'- {review}')
```

### Post-Cleaning Duplicate Verification

Although duplicates were removed earlier, this step re-verifies that no duplicate reviews exist within the training or testing sets after all cleaning and processing. This ensures the integrity of the data used for model training and evaluation.

```python
print('\n--- Post-Cleaning Duplicate Verification ---')
duplicates_in_X_train = X_train_str.duplicated().sum()
duplicates_in_X_test = X_test_str.duplicated().sum()

print(f'Number of duplicate reviews within X_train (after cleaning): {duplicates_in_X_train}')
print(f'Number of duplicate reviews within X_test (after cleaning): {duplicates_in_X_test}')

if duplicates_in_X_train == 0 and duplicates_in_X_test == 0:
    print('No duplicates found within training or testing sets after cleaning.')
else:
    print('Warning: Duplicates found within training or testing sets after cleaning. This might indicate an issue with prior deduplication steps or that the duplicates were introduced during cleaning (unlikely for this specific cleaning).')
```

### TF-IDF Feature Engineering

TF-IDF (Term Frequency-Inverse Document Frequency) is a numerical statistic that is intended to reflect how important a word is to a document in a collection or corpus. It is often used as a weighting factor in information retrieval and text mining. This step transforms the text data into numerical features that can be used by machine learning models.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

print('\n--- TF-IDF Feature Engineering ---')
# Initialize TfidfVectorizer
# Using a reasonable max_features to control vocabulary size
tfidf_vectorizer = TfidfVectorizer(max_features=10000)

# Fit the vectorizer on the training data and transform both training and testing data
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train_str)
X_test_tfidf = tfidf_vectorizer.transform(X_test_str)

print(f'Shape of X_train_tfidf: {X_train_tfidf.shape}')
print(f'Shape of X_test_tfidf: {X_test_tfidf.shape}')
```

### Saving TF-IDF Matrices

Saving the TF-IDF transformed matrices (`X_train_tfidf` and `X_test_tfidf`) allows for quick loading and reuse in subsequent modeling steps without needing to re-run the entire TF-IDF transformation process. This is particularly useful for large datasets.

```python
from scipy.sparse import save_npz

print('\n--- Saving TF-IDF Matrices ---')
save_npz('X_train_tfidf.npz', X_train_tfidf)
save_npz('X_test_tfidf.npz', X_test_tfidf)
print('X_train_tfidf.npz and X_test_tfidf.npz saved.')
```

### Saving the TF-IDF Vectorizer

Saving the `TfidfVectorizer` object itself is crucial. It contains the learned vocabulary and inverse document frequencies. This allows for consistent transformation of new, unseen text data (e.g., during prediction) using the same logic applied to the training data.

```python
import pickle

print('\n--- Saving TF-IDF Vectorizer ---')
with open('tfidf_vectorizer.pkl', 'wb') as f:
    pickle.dump(tfidf_vectorizer, f)
print('tfidf_vectorizer.pkl saved.')
```

### Saving Feature Names

The feature names (i.e., the words corresponding to each column in the TF-IDF matrix) are useful for interpretability, allowing us to understand which words contribute most to predictions. Saving them separately makes it easy to access this information later.

```python
import numpy as np

print('\n--- Saving Feature Names ---')
feature_names = tfidf_vectorizer.get_feature_names_out()
np.save('tfidf_feature_names.npy', feature_names)
print('tfidf_feature_names.npy saved.')
```

### Data Quality Audit Summary

This final section provides a concise summary of the data quality checks performed, including information on potential leakage, duplicate presence, and the outcome of the feature engineering step. This serves as a quick overview of the dataset's readiness for modeling.

### Removing Duplicates from Training Set

Although a general duplicate check was performed, specific duplicates within the `X_train` set were identified. This step removes those duplicates to ensure that each sample in the training data is unique, preventing potential biases in model training.

```python
print('--- Removing duplicates from X_train ---')

# Identify duplicate indices in X_train_str
duplicate_indices_train = X_train_str[X_train_str.duplicated()].index

if not duplicate_indices_train.empty:
    print(f'Found {len(duplicate_indices_train)} duplicate entries in X_train. Removing them...')

    # Drop duplicates from X_train_str, X_train, and y_train
    X_train_str = X_train_str.drop(index=duplicate_indices_train)
    X_train = X_train.drop(index=duplicate_indices_train)
    y_train = y_train.drop(index=duplicate_indices_train)

    print('Duplicates removed successfully.')
    print(f'New X_train length: {len(X_train)} samples')
    print(f'New y_train length: {len(y_train)} samples')
else:
    print('No duplicates found in X_train to remove.')
```

```python
# ============================================================
# REMOVE DUPLICATES FROM TRAINING SET
# ============================================================

duplicate_indices_train = X_train_str[
    X_train_str.duplicated()
].index

print(
    f"Duplicate training samples found: "
    f"{len(duplicate_indices_train)}"
)

if len(duplicate_indices_train) > 0:

    X_train = X_train.drop(
        index=duplicate_indices_train
    )

    X_train_str = X_train_str.drop(
        index=duplicate_indices_train
    )

    y_train = y_train.drop(
        index=duplicate_indices_train
    )

    print("✓ Training duplicates removed.")

else:

    print("✓ No training duplicates found.")


print("\nFinal training samples:")
print(len(X_train))
```

```python
# ============================================================
# TRAIN / TEST LEAKAGE CHECK AND REMOVAL
# ============================================================

print("=" * 60)
print("TRAIN / TEST LEAKAGE CHECK")
print("=" * 60)

train_reviews = set(X_train_str)

common_reviews = (
    train_reviews.intersection(
        set(X_test_str)
    )
)

print(
    f"Common reviews before removal: "
    f"{len(common_reviews)}"
)


# ------------------------------------------------------------
# Remove overlapping reviews from TEST set
# ------------------------------------------------------------

if len(common_reviews) > 0:

    overlap_mask = X_test_str.isin(common_reviews)

    X_test = X_test[
        ~overlap_mask
    ]

    X_test_str = X_test_str[
        ~overlap_mask
    ]

    y_test = y_test[
        ~overlap_mask
    ]

    print(
        f"✓ Removed {overlap_mask.sum()} "
        f"overlapping review(s) from test set."
    )

else:

    print("✓ No overlapping reviews found.")


# ------------------------------------------------------------
# Verify again
# ------------------------------------------------------------

remaining_overlap = (
    set(X_train_str)
    .intersection(set(X_test_str))
)

print(
    f"\nRemaining train/test overlap: "
    f"{len(remaining_overlap)}"
)

if len(remaining_overlap) == 0:
    print("✓ No train/test leakage remains.")
else:
    print("⚠️ Leakage still exists.")
```
